# 0. Download the dataset

```python
import os, zipfile, urllib.request

os.makedirs("data", exist_ok=True)
url = "https://raw.githubusercontent.com/MWiechmann/enron_spam_data/master/enron_spam_data.zip"
zpath, cpath = "data/enron_spam_data.zip", "data/enron_spam_data.csv"

if not os.path.exists(cpath):
    if not os.path.exists(zpath):
        urllib.request.urlretrieve(url, zpath)
    zipfile.ZipFile(zpath).extractall("data")

print("Working folder:", os.getcwd())
print("Files:", os.listdir("data"))
print("CSV size (MB):", round(os.path.getsize(cpath) / 1e6, 1))
```

# 1. Import Libraries


```python
import json, re, time, warnings
import numpy as np, pandas as pd
import sklearn, xgboost
import joblib
from sklearn.compose import ColumnTransformer
from sklearn.ensemble import RandomForestClassifier
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (accuracy_score, confusion_matrix, f1_score, precision_score,
                             recall_score, roc_auc_score, classification_report)
from sklearn.model_selection import StratifiedKFold, cross_val_score, train_test_split
from sklearn.neural_network import MLPClassifier
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeClassifier
from xgboost import XGBClassifier

warnings.filterwarnings("ignore")
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 250)
pd.set_option("display.max_colwidth", 90)
SEED = 42
print("pandas", pd.__version__, "| scikit-learn", sklearn.__version__, "| xgboost", xgboost.__version__)
```

# 2. Load the raw data


```python
raw = pd.read_csv("data/enron_spam_data.csv")
raw.columns = ["id", "subject", "message", "label", "date"]
print("Shape (rows, columns):", raw.shape)
raw.head(5)
```

```python
print("Missing values per column:")
print(raw.isna().sum())
print("\nLabel counts:")
raw.label.value_counts()
```

# 3. Clean the data


```python
df = raw.copy()
df["subject"] = df.subject.fillna("")
df["message"] = df.message.fillna("")
df["text_raw"] = (df.subject + " . " + df.message).str.strip()

before = len(df)
df = df[df.text_raw.str.len() > 5].drop_duplicates(subset="text_raw").reset_index(drop=True)
print(f"Rows before: {before} | after cleaning: {len(df)} | removed: {before - len(df)}")
df[["subject", "label", "text_raw"]].head(5)
```

# 4. Email label to email type ("attack name to attack family")


```python
URL_RE = re.compile(r"https?://\S+|www\.\S+", re.I)
PHISH_RE = re.compile(r"\b(account|password|passwd|verify|verification|login|log in|sign in|bank|paypal|ebay|"
                      r"credit card|debit card|billing|suspended|security alert|confirm your|ssn|social security)\b", re.I)
CLICK_RE = re.compile(r"\bclick\b|https?://|www\.", re.I)

def email_family(label, text):
    if label == "ham":
        return "Ham"
    return "Phishing" if (PHISH_RE.search(text) and CLICK_RE.search(text)) else "Spam"

tests = [("ham", "Meeting moved to 3pm tomorrow."),
         ("spam", "Buy cheap pills now, best price guaranteed!!!"),
         ("spam", "Your account is suspended. Click http://bit.ly/x to verify your password.")]
pd.DataFrame([(l, t[:60], email_family(l, t)) for l, t in tests], columns=["label", "text", "family"])
```

# 5. Add the answers to columns


```python
df["family"] = [email_family(l, t) for l, t in zip(df.label, df.text_raw)]
df["attack"] = (df.label != "ham").astype(int)
assert df.family.notna().all()
print("Family counts:")
print(df.family.value_counts())
df[["subject", "label", "family", "attack"]].head(8)
```

```python
print("Share of each family (%):")
(df.family.value_counts(normalize=True) * 100).round(2)
```

# 6. Train / test split


```python
tr, te = train_test_split(df, test_size=0.2, stratify=df.family, random_state=SEED)
tr, te = tr.reset_index(drop=True), te.reset_index(drop=True)
print("Train shape:", tr.shape, "| Test shape:", te.shape)
pd.DataFrame({"train": tr.family.value_counts(), "test": te.family.value_counts()})
```

# 7. Feature engineering (the email version of "connection features")


```python
SHORTENERS = ("bit.ly", "tinyurl", "goo.gl", "t.co", "ow.ly", "is.gd")
URGENT = ("urgent", "verify", "suspended", "immediately", "confirm", "password", "winner", "prize", "free",
          "click", "limited time", "act now", "refund", "security alert", "congratulations", "guaranteed")

def make_numeric(d):
    t = d.text_raw
    low = t.str.lower()
    urls = t.apply(lambda s: URL_RE.findall(s))
    letters = t.str.count(r"[A-Za-z]").clip(lower=1)
    out = pd.DataFrame({
        "num_chars":     t.str.len(),
        "num_words":     t.str.split().str.len(),
        "subject_len":   d.subject.str.len(),
        "num_urls":      urls.str.len(),
        "has_shortener": urls.apply(lambda L: int(any(s in u.lower() for u in L for s in SHORTENERS))),
        "has_ip_url":    urls.apply(lambda L: int(any(re.match(r"https?://\d+\.\d+\.\d+\.\d+", u) for u in L))),
        "url_has_at":    urls.apply(lambda L: int(any("@" in u for u in L))),
        "num_exclaim":   t.str.count("!"),
        "num_dollar":    t.str.count(r"\$"),
        "upper_ratio":   t.str.count(r"[A-Z]") / letters,
        "digit_ratio":   t.str.count(r"\d") / t.str.len().clip(lower=1),
        "urgent_words":  low.apply(lambda s: sum(s.count(w) for w in URGENT)),
        "has_html":      low.str.contains(r"<html|<a href|<br>", regex=True).astype(int),
    }, index=d.index)
    return out

num = list(make_numeric(tr.head(2)).columns)
for d in (tr, te):
    d[num] = make_numeric(d)
print("Numeric features:", num)
tr[["family"] + num].head(5).round(3)
```

```python
print("Average of each numeric feature per family (do the families really look different?):")
tr.groupby("family")[num].mean().round(3).T
```

# 8. Clean the text for the model


```python
def clean_text(s):
    s = re.sub(r"<[^>]+>", " ", s)
    s = URL_RE.sub(" urltoken ", s)
    s = re.sub(r"[^a-zA-Z\s]", " ", s).lower()
    return re.sub(r"\s+", " ", s).strip()

for d in (tr, te):
    d["clean_text"] = d.text_raw.apply(clean_text)

pd.DataFrame({"family": tr.family.head(3),
              "original": tr.text_raw.head(3).str[:110],
              "cleaned": tr.clean_text.head(3).str[:110]})
```

# 9. Separate features from answers


```python
feats = ["clean_text"] + num
Xtr, Xte = tr[feats], te[feats]
print("Input columns:", feats)
print("Xtr:", Xtr.shape, "| Xte:", Xte.shape)
Xtr.head(5)
```

# 10. Summary facts


```python
R = {"train_rows": len(tr), "test_rows": len(te), "n_numeric_features": len(num),
     "train_attack_share": float(tr.attack.mean()), "test_attack_share": float(te.attack.mean()),
     "train_family_counts": tr.family.value_counts().to_dict(),
     "test_family_counts": te.family.value_counts().to_dict()}
print(json.dumps(R, indent=1))
```

# 11. Preprocessing (TF-IDF for text, scaling for numbers)


```python
def prep():
    return ColumnTransformer([
        ("text", TfidfVectorizer(max_features=3000, ngram_range=(1, 2), stop_words="english", min_df=5), "clean_text"),
        ("num", StandardScaler(), num)])

demo = prep()
clean = demo.fit_transform(Xtr)
names = demo.get_feature_names_out()
print("Before:", Xtr.shape, "-> After:", clean.shape, "(3000 TF-IDF terms + %d numeric columns)" % len(num))
print("Sample feature names:", list(names[:5]), "...", list(names[-3:]))
```

## TF-IDF: which words describe the first email?


```python
row = clean[0]
row = row.toarray().ravel() if hasattr(row, "toarray") else np.asarray(row).ravel()
text_cols = [i for i, n in enumerate(names) if n.startswith("text__")]
top = sorted(text_cols, key=lambda i: row[i], reverse=True)[:8]
print("Email family:", tr.family[0])
print("Email start :", tr.text_raw[0][:120].replace("\n", " "))
pd.DataFrame({"word": [names[i].replace("text__", "") for i in top], "tfidf_weight": [round(row[i], 3) for i in top]})
```

## Scaling: before and after


```python
cols4 = ["num_chars", "num_urls", "urgent_words", "upper_ratio"]
print("BEFORE scaling:")
display(Xtr[cols4].head(5).round(3))
dense = clean.toarray() if hasattr(clean, "toarray") else clean
clean_df = pd.DataFrame(dense[:, -len(num):], columns=["num__" + c for c in num])
print("AFTER scaling:")
display(clean_df[["num__" + c for c in cols4]].head(5).round(3))
print("Mean / std after scaling:")
clean_df[["num__" + c for c in cols4]].agg(["mean", "std"]).round(3)
```

# Define the models

```python
models = {
    "Logistic Regression": LogisticRegression(max_iter=2000),
    "Decision Tree": DecisionTreeClassifier(max_depth=20, random_state=SEED),
    "Random Forest": RandomForestClassifier(100, n_jobs=-1, random_state=SEED),
    "Neural Network (MLP)": MLPClassifier((64, 32), early_stopping=True, max_iter=100, random_state=SEED),
    "XGBoost": XGBClassifier(n_estimators=200, max_depth=6, learning_rate=0.1, subsample=0.8,
                             colsample_bytree=0.8, eval_metric="logloss", n_jobs=-1, random_state=SEED),
}
print("Models:", list(models))
```

# 13. Train and test every model (binary: legitimate vs spam/phishing)


```python
rows, fitted = [], {}
for name, m in models.items():
    pipe = make_pipeline(prep(), m)
    cv = cross_val_score(pipe, Xtr, tr.attack, cv=StratifiedKFold(3, shuffle=True, random_state=SEED), scoring="f1")
    t = time.time(); pipe.fit(Xtr, tr.attack); fit = time.time() - t
    p = pipe.predict(Xte); s = pipe.predict_proba(Xte)[:, 1]
    tn, fp, fn, tp = confusion_matrix(te.attack, p).ravel()
    row = {"model": name, "cv_f1": cv.mean(), "cv_f1_std": cv.std(), "accuracy": accuracy_score(te.attack, p),
           "precision": precision_score(te.attack, p), "recall": recall_score(te.attack, p),
           "f1": f1_score(te.attack, p), "fpr": fp / (fp + tn), "roc_auc": roc_auc_score(te.attack, s),
           "tp": int(tp), "fp": int(fp), "tn": int(tn), "fn": int(fn), "train_s": fit}
    rows.append(row); fitted[name] = pipe
    print({k: (round(v, 4) if isinstance(v, float) else v) for k, v in row.items()}, flush=True)
R["binary_models"] = rows
```

```python
pd.DataFrame(rows)[["model", "cv_f1", "accuracy", "precision", "recall", "f1", "fpr", "roc_auc", "train_s"]].round(4)
```

## Look at one model's predictions (XGBoost)


```python
pipe = fitted["XGBoost"]
p = pipe.predict(Xte); s = pipe.predict_proba(Xte)[:, 1]
res = pd.DataFrame({"subject": te.subject.str[:45], "family": te.family, "truth": te.attack,
                    "guess": p, "attack_prob": s.round(3)})
res.head(10)
```

```python
wrong = res[res.truth != res.guess]
print("Total mistakes:", len(wrong), "out of", len(res))
wrong.head(10)
```

```python
tn, fp, fn, tp = confusion_matrix(te.attack, p).ravel()
pd.DataFrame([[tn, fp], [fn, tp]], index=["really legitimate", "really spam/phishing"],
             columns=["says legitimate", "says spam/phishing"])
```

```python
res["correct"] = (res.truth == res.guess)
res.groupby("family").correct.mean().round(3)
```

# 14. Three-class model (Ham / Phishing / Spam)


```python
best = max(rows, key=lambda r: r["cv_f1"])["model"]
mc = make_pipeline(prep(), type(models[best])(**models[best].get_params()))
labels = sorted(tr.family.unique()); lab = {l: i for i, l in enumerate(labels)}
inv = {i: l for l, i in lab.items()}
print("Best model by CV F1:", best)
print("Label mapping:", lab)
```

```python
mc.fit(Xtr, tr.family.map(lab)); pm = mc.predict(Xte)
pd.DataFrame({"subject": te.subject.str[:45], "true_family": te.family, "guess": [inv[i] for i in pm]}).head(10)
```

```python
cm3 = confusion_matrix(te.family.map(lab), pm)
pd.DataFrame(cm3, index=["really " + l for l in labels], columns=["says " + l for l in labels])
```

```python
rep = classification_report(te.family.map(lab), pm, target_names=labels, output_dict=True, zero_division=0)
pd.DataFrame(rep).T.round(3)
```

# 15. Try it on new emails

```python
final_binary = fitted[best]

def check_email(subject, body):
    d = pd.DataFrame({"subject": [subject], "message": [body]})
    d["text_raw"] = (d.subject + " . " + d.message).str.strip()
    d[num] = make_numeric(d)
    d["clean_text"] = d.text_raw.apply(clean_text)
    X = d[feats]
    prob = float(final_binary.predict_proba(X)[0, 1])
    fam = inv[int(mc.predict(X)[0])]
    return {"subject": subject[:40], "risk_score": round(prob, 3),
            "verdict": "SPAM/PHISHING" if prob >= 0.5 else "LEGIT", "type": fam}

samples = [
    ("Meeting moved to 3pm", "Hi all, the project meeting is moved to 3pm tomorrow. Please review the attached agenda."),
    ("URGENT: verify your account", "Your bank account is suspended. Click http://bit.ly/x1 now and confirm your password immediately!"),
    ("You are a WINNER!!!", "Congratulations, you won $1,000,000. Act now, limited time, 100% guaranteed, buy cheap pills!!!"),
    ("Lunch on Friday?", "Are you free for lunch on Friday? Let me know what time works for you."),
]
pd.DataFrame([check_email(s, b) for s, b in samples])
```

# 16. Save results

```python
R["multiclass"] = {"model": best, "accuracy": accuracy_score(te.family.map(lab), pm),
                   "macro_f1": f1_score(te.family.map(lab), pm, average="macro"),
                   "per_class": {k: {m: round(v, 4) for m, v in rep[k].items()} for k in labels}}
R["selected_by_cv_f1"] = best
print(json.dumps(R["multiclass"], indent=1))
json.dump(R, open("results.json", "w"), indent=1, default=float)
joblib.dump(final_binary, "email_binary_model.joblib")
joblib.dump({"model": mc, "labels": inv}, "email_family_model.joblib")
print("\nDONE. Saved:", [f for f in os.listdir(".") if f.endswith((".json", ".joblib"))])
```
